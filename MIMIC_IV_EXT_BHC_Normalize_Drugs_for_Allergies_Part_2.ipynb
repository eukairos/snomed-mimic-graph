This notebook continues cleaning the dataset of MIMIC-IV drug allergies.  
Here we partition the full set of allergies from the last notebook into those that were matched, and those that were not matched, into 2 separate csv files for further processing.
In production, of course, each addition or omission can be a life-and-death matter, so your clinical partners MUST review every single entry manually. NO EXCEPTIONS.

```python
import pandas as pd
data = pd.read_csv('./data/MIMIC_IV_ALLERGIEX_RXNORM_FULL.csv')
data
```

```python
fully_matched = data.dropna(subset=['matched_name']).reset_index(drop=True)
fully_matched
```

```python
fully_matched.to_csv('./data/MIMIC_IV_EXT_BHC_Allergies_Neo4J_Fully_Matched.csv', index=False)
```

```python
unmatched = data[data['matched_name'].isna()].reset_index(drop=True)
unmatched
```

```python
unmatched.to_csv('./data/MIMIC_IV_EXT_BHC_Allergies_Neo4J_Unmatched.csv', index=False)
```
