### Cleaning script for SNOMED files before loading into neo4j

First, sign up with the US National Library of Medicine for a UMLS account. Navigate to the SNOMED CT International Edition page: https://www.nlm.nih.gov/healthit/snomedct/international.html and download the RF2 files. The files are large, may take a while to download.  
Do remember to use the files only for your academic/research work, so as to avoid user licensing violations. If you are intending to use SNOMED for commercial purposes, please obtain a commercial license from SNOMED.

There are 2 versions of SNOMED: full and snapshot. For our demo purposes, we'll use the simpler snapshot version. In production environments, of course, it will be more robust to implement the full version.  

Extract the downloaded zip file, and open the snapshot folder. There are 2 subfolders, Refset and Terminology. Open the terminology folder. The 3 files we're interested in are the sct2_Concept_Snapshot_INT_20XXXXXX.txt, sct2_Description_....txt, and sct2_Relationship_.....txt. Copy these files into your working directory.  I renamed these files as concept.txt, description.txt and relationship.txt for convenience. 


```python
import pandas as pd
concept = pd.read_csv('concept.txt', sep='\t')
desc = pd.read_csv('description.txt', sep='\t')
relate = pd.read_csv('relationship.txt', sep='\t')
```

```python
# For our purposes, we only need the active concepts
concept = concept[concept['active']==1]
desc = desc[desc['active']==1]
relate = relate[relate['active'] == 1]
# we can now drop the active column
```

```python
# remove unnecessary columns 
for item in ['effectiveTime', 'active']:
    concept = concept.drop(item, axis=1)
concept
```

```python
# quotation marks and some fonts within the text of the term attribute of the description file cause laoding errors, we'll remove them. 
to_clean = ['"','©']
for item in to_clean:
    desc['term'] = desc['term'].str.replace(item, '')
```

```python
# inspect the files
concept
```

```python
desc = desc.drop('active', axis=1)
desc
```

```python
relate = relate.drop(['active', 'effectiveTime'], axis=1)
relate
```

```python
concept.to_csv('snomed_concept.csv', index=False)
desc.to_csv('snomed_description.csv', index=False)
relate.to_csv('snomed_relationship.csv', index=False)
```
